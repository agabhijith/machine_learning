```python
# 1. Import the required libraries.
# 2. Load and clean the Online Retail dataset.
# 3. Calculate the total amount for each transaction.
# 4. Group transactions by customer and create customer features.
# 5. Divide customers into Low, Medium, and High Value segments.
# 6. Split the data into training and testing sets.
# 7. Train an entropy-based Decision Tree classifier.
# 8. Predict customer segments and evaluate the model.
# 9. Display the classification report and confusion matrix.
# 10. Plot the decision tree and feature importance.
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.metrics import *



df = pd.read_excel('Online Retail.xlsx')
df = df.dropna(subset=['CustomerID'])
df = df[~df.InvoiceNo.astype(str).str.startswith('C')]
df = df[(df.Quantity > 0) & (df.UnitPrice > 0)]
df['TotalAmount'] = df.Quantity * df.UnitPrice

c = df.groupby('CustomerID').agg(
    TotalSpend=('TotalAmount', 'sum'),
    TotalQuantity=('Quantity', 'sum'),
    Transactions=('InvoiceNo', 'nunique'),
    Products=('StockCode', 'nunique'),
    AverageOrder=('TotalAmount', 'mean'))

y = pd.qcut(c.TotalSpend, 3, labels=['Low', 'Medium', 'High'])
X = c.drop(columns='TotalSpend')

a, b, d, e = train_test_split(X, y, test_size=0.2, stratify=y, random_state=42)
m = DecisionTreeClassifier(criterion='entropy', max_depth=4, random_state=42)
m.fit(a, d)
p = m.predict(b)

print("Accuracy:", accuracy_score(e, p))
print(classification_report(e, p))
print(confusion_matrix(e, p))

plot_tree(m, feature_names=X.columns, class_names=m.classes_, filled=True)
plt.show()

pd.Series(m.feature_importances_, index=X.columns).plot.bar()
plt.show()
```
